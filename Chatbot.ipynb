```python
def show_menu():
    print("\n--- Customer Support Menu ---")
    print("1. Store Hours")
    print("2. Location")
    print("3. Return Policy")
    print("4. Contact Info")
    print("5. Say Hello")
    print("6. Exit")
    print("-----------------------------")

print("Welcome to ShopEase Customer Support!")
show_menu()

while True:
    user_input = input("\nYou: ").lower()

    if user_input in ["1", "store hours", "timing"]:
        print("Bot: Our store is open from 9 AM to 9 PM, Monday to Saturday.")
    elif user_input in ["2", "location", "address"]:
        print("Bot: We are located at MG Road, Nashik, near City Mall.")
    elif user_input in ["3", "return policy"]:
        print("Bot: You can return products within 7 days with a valid receipt.")
    elif user_input in ["4", "contact", "phone"]:
        print("Bot: You can reach us at +91-9876543210.")
    elif user_input in ["5", "hello", "hi"]:
        print("Bot: Hello! How can I help you today?")
    elif user_input in ["6", "exit", "bye"]:
        print("Bot: Thank you for contacting ShopEase. Have a great day!")
        break
    else:
        print("Bot: I'm sorry, I didn't understand that. Please enter a valid option.")

    show_menu()
```
