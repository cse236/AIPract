```python
edges = [
    (2, 'A', 'B'),
    (3, 'A', 'C'),
    (1, 'B', 'C'),
    (1, 'B', 'D'),
    (4, 'C', 'D')
]
 
parent = {}
 
def find(node):
    if parent[node] != node:
        parent[node] = find(parent[node])
    return parent[node]
 
def union(u, v):
    parent[find(u)] = find(v)
 
def kruskal(edges):
    for node in ['A', 'B', 'C', 'D']:
        parent[node] = node
 
    edges.sort()
    total_cost = 0
 
    for weight, u, v in edges:
        if find(u) != find(v):
            print(f"Edge selected: {u} - {v} with weight {weight}")
            union(u, v)
            total_cost += weight
 
    print("Total Minimum Cost:", total_cost)
 
kruskal(edges)
```
